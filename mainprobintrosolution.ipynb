## Challenge 1
There is a box with 10 white balls, 12 red balls and 8 black balls. Calculate the probability of:
* Taking a white ball out.
* Taking a white ball on first extraction and a black ball out on second extraction without replacing the first one.
* Taking a white ball on first extraction and a black ball out on second extraction replacing the first one.
* Taking 3 balls and them being all different colours.

```python
from math import comb
total_balls = 10 + 12 + 8
white_balls = 10
red_balls = 12
black_balls = 8

# Taking the white ball out:
prob_white = white_balls / total_balls

# Taking a white ball on first extraction and a black ball out on second extraction without replacing the first one.
prob_white_first = white_balls / total_balls
prob_black_second = black_balls / (total_balls - 1)
prob_white_then_black_no_replace = prob_white_first * prob_black_second


# Taking a white ball on first extraction and a black ball out on second extraction replacing the first one.
prob_black_second_replace = black_balls / total_balls
prob_white_then_black_replace = prob_white_first * prob_black_second_replace


# Taking 3 balls and them being all different colours.
# Ways to choose one of each color
ways_white_red_black = comb(white_balls, 1) * comb(red_balls, 1) * comb(black_balls, 1)
# Total ways to choose any 3 balls
ways_total_3_balls = comb(total_balls, 3)
prob_all_different = ways_white_red_black / ways_total_3_balls

```

## Challenge 2

You want to go on picnic on Saturday. You hate rain, and would know to plan your weekend. To help you make a decision, you gather the following data about rainy days:

* 40% of all days are cloudy.
* About 50% of cloudy days are rainy. 
* About 20% of non-cloudy days are rainy.

What is the chance of rain during next Saturday?

```python

p_cloudy = 0.4 
p_rain_given_cloudy = 0.5  
p_not_cloudy = 1 - p_cloudy  
p_rain_given_not_cloudy = 0.2  

# Using the law of total probability
p_rain = (p_cloudy * p_rain_given_cloudy) + (p_not_cloudy * p_rain_given_not_cloudy)
```

## Challenge 3

A and B are events of a probability such that $P(A) = 0.3$, $P(B) = 0.6$ and $P(A \cap B) = 0.1$

Which of the following statements are true?  
**HINT: draw a big Venn Diagram by your side**  
* $P(\overline{A}) = 0.7$
* $P(A \cup B) = 0.6$
* $P(A \cap \overline{B}) = 0.2$
* $P(A \cup \overline{A}) = 1$  
* $P(A \cap (B \cup \overline{B})) = 0.4$
* $P(\overline{A} \cap \overline{B}) = 0.3$
* $P(\overline{A \cap B}) = 0.9$

```python
# Given probabilities
P_A = 0.3
P_B = 0.6
P_A_and_B = 0.1

# 1.
P_not_A = 1 - P_A
print(f"1. P(¬A) = {P_not_A:.1f}, True: {P_not_A == 0.7}")

# 2.
P_A_or_B = P_A + P_B - P_A_and_B
print(f"2. P(A ∪ B) = {P_A_or_B:.1f}, True: {P_A_or_B == 0.6}")

# 3. 
P_A_and_not_B = P_A - P_A_and_B
print(f"3. P(A ∩ ¬B) = {P_A_and_not_B:.1f}, True: {P_A_and_not_B == 0.2}")

# 4. 
P_A_or_not_A = 1
print(f"4. P(A ∪ ¬A) = {P_A_or_not_A:.1f}, True: {P_A_or_not_A == 1}")

# 5. 
P_A_and_B_or_not_B = P_A
print(f"5. P(A ∩ (B ∪ ¬B)) = {P_A_and_B_or_not_B:.1f}, True: {P_A_and_B_or_not_B == 0.4}")

# 6. 
P_not_A_and_not_B = 1 - P_A_or_B
print(f"6. P(¬A ∩ ¬B) = {P_not_A_and_not_B:.1f}, True: {P_not_A_and_not_B == 0.3}")

# 7. 
P_not_A_and_B = 1 - P_A_and_B
print(f"7. P(¬(A ∩ B)) = {P_not_A_and_B:.1f}, True: {P_not_A_and_B == 0.9}")
```

## Challenge 4

 * If someone flies to China, the probability of getting Covid is 1%  
 * If someone flies to Italy, the probability of getting Covid is 20%  

I roll 1 dice with numbers 1, 2, 3, 4, 5, 6:  
 * If 1: I fly to Italy
 * If not 1: I fly to China

During my trip, I got Covid. What is the probability that I flew to China?

```python

P_Italy = 1 / 6  
P_China = 1 - P_Italy 

P_Covid_given_Italy = 0.2  
P_Covid_given_China = 0.01 


P_Covid = (P_Covid_given_Italy * P_Italy) + (P_Covid_given_China * P_China)


P_China_given_Covid = (P_Covid_given_China * P_China) / P_Covid

print(f"The probability that I flew to China given I got Covid is: {P_China_given_Covid:.4f}")
```

## BONUS

You roll 2 dice. What is the probability that the sum of the numbers is 7? And 8? And 9? Then simulate 10000 times and draw a countplot.

your explanation here

```python
import matplotlib.pyplot as plt
import seaborn as sns
import random


total_outcomes = 6 * 6


favorable_7 = sum(1 for i in range(1, 7) for j in range(1, 7) if i + j == 7)
favorable_8 = sum(1 for i in range(1, 7) for j in range(1, 7) if i + j == 8)
favorable_9 = sum(1 for i in range(1, 7) for j in range(1, 7) if i + j == 9)


prob_7 = favorable_7 / total_outcomes
prob_8 = favorable_8 / total_outcomes
prob_9 = favorable_9 / total_outcomes

print(f"Probability of sum 7: {prob_7:.4f}")
print(f"Probability of sum 8: {prob_8:.4f}")
print(f"Probability of sum 9: {prob_9:.4f}")


simulated_sums = [random.randint(1, 6) + random.randint(1, 6) for _ in range(10000)]


sns.countplot(x=simulated_sums, order=range(2, 13), color="skyblue")
plt.title("Count of Sums from 10,000 Dice Rolls")
plt.xlabel("Sum")
plt.ylabel("Count")
plt.show()
```
